```python
#Remover linhas com valores ausentes
df_dados.dropna(inplace=True) #comando mais simples, exlui todas as linhas que tem 1 valor vazio
display(df_dados)
print(df_dados.isnull().sum())
```

```python
#Selecionar linhas com valores não-negativos
#Selecionando as linhas onde os valores das colunas
# “Dias desde a ultima Atualizacao” e “Classificacao” são maiores ou iguais a zero.

mascara_booleana = (df_dados['Dias desde a ultima Atualizacao'] >= 0) & (df_dados['Classificacao'] >= 0)

#Atualiza o DataFrame para manter apenas as linhas selecionadas pela máscara
df_dados = df_dados[mascara_booleana]

print(df_dados.shape)

print(df_dados.shape[0])
```

```python
#Split dos Dados em Subconjuntos de Treinamento e Teste
#dividindo os dados em dois subconjuntos: um para treinamento e outro para teste

from sklearn.model_selection import train_test_split

#Selecione todas as colunas do DataFrame df, exceto a coluna “No de Reviews",
#e atribua essas features a uma variável chamada X.

#Fica com todas as colunas menos a Classificação (as pistas / features).
x = df_dados.drop('Classificacao', axis=1)

#Fica somente com a coluna Classificação (a resposta / alvo).
y = df_dados['Classificacao']

#Dividindo os dados em subconjuntos de treinamento e teste
x_treinamento, x_teste, y_treinamento, y_teste = train_test_split(x, y, test_size=0.2, random_state=0)


print("Tamanho do Treinamento (X):", x_treinamento.shape)
print("Tamanho do Teste (X):", x_teste.shape)
print("Tamanho do Treinamento (y):", y_treinamento.shape)
print("Tamanho do Teste (y):", y_teste.shape)
```

```python
#Verificando se há Correlação forte entre as variáveis independentes
#Verificando a correlação entre as features numéricas do conjunto de dados de treinamento e 
#identificando possíveis pares de features com alta correlação.

#Removendo a coluna “Categoria” do DataFrame X_treinamento, uma vez que ela contém valores categóricos
# e não será incluída no cálculo da matriz de correlação.
#Convertendo as colunas relevantes do DataFrame X_treinamento_numeric para tipos numéricos usando a função pd.to_numeric(). 
# As colunas a serem convertidas são: ‘No de Reviews', ‘No de Instalacoes', ‘Tamanho', ‘Preco' e ‘Dias desde a ultima Atualizacao'. 
# Utiliando o parâmetro errors='coerce' para lidar com possíveis erros de conversão.

x_treinamento_numerico = x_treinamento.drop('Categoria', axis=1)

colunas_numericas = ['No de Reviews', 'No de Instalacoes', 'Tamanho', 'Preco', 'Dias desde a ultima Atualizacao']
x_treinamento_numerico[colunas_numericas] = x_treinamento_numerico[colunas_numericas].apply(pd.to_numeric, errors='coerce') #Pandas vai passar por cada uma das colunas listadas e aplicar a conversão forçando o errors='coerce' onde for necessário.

matriz_correlacao = x_treinamento_numerico.corr() #calcula a correlação (geralmente usando a correlação de Pearson por padrão) entre todas as combinações possíveis de colunas
print(matriz_correlacao)
```

```python
#Criando Gráficos de Dispersão

# Lista das colunas numéricas sugeridas
colunas_plot = ['No de Reviews', 'No de Instalacoes', 'Tamanho', 'Preco', 'Dias desde a ultima Atualizacao']

#Criando a figura com 2 linhas e 3 colunas e definindo o tamanho (16, 10)
fig, axes = plt.subplots(nrows=2,ncols=3,figsize=(16,10))
fig.suptitle('Graficos de Dispersao: Classificaçao vs Outras Colunas', fontsize=16)  #titulo superior
fig.tight_layout(pad=5.0)  #layout, ajusta espacamento entre os subplotes, evitando sopreposisao

#loop percorrendo colunas a serem plotadas e indices

for i, column in enumerate(colunas_plot):
    row = i // 3 #resto da divisao por 3
    col = i % 3 #modulo
    axes[row, col].scatter(x_treinamento_numerico[column], y_treinamento)
    axes[row, col].set_title(f'Classificacao vs {column}')
    axes[row, col].set_xlabel(column)
    axes[row, col].set_ylabel('Classificaçao')

axes[1, 2].set_visible(False) #esconde a coluna 6, POIS ESTA VAZIA
plt.show()
```

```python
#imputando a média em Tamanho (SimpleImputer), padronização das colunas numéricas (StandardScaler) e 
# codificação one-hot de Categoria (OneHotEncoder)

from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

#Declarar as variaveis num_col e cat_col
num_col = ['No de Reviews', 'No de Instalacoes', 'Tamanho', 'Preco', 'Dias desde a ultima Atualizacao']
cat_col = ['Categoria']

#Substituindo os valores ausentes na coluna 'Tamanho' pela média
imp = SimpleImputer(strategy='mean')
tf_num = imp.fit_transform(x_treinamento[num_col])

#Escalando as colunas numericas, atribuindo array, tf num recebe
scaler = StandardScaler()
tf_num = scaler.fit_transform(tf_num)

#instanciando um objeto
#codificando a coluna 'Categoria' usando a codificaçao one-hot
ohe = OneHotEncoder(sparse_output=False, drop='first')
tf_cat = ohe.fit_transform(x_treinamento[cat_col])

#Concatenando os arrays tf_num e tf_cat ao longo do eixo 1
x_treinamento_transformado = np.concatenate((tf_num, tf_cat), axis=1)

print(f'Exemplo transformado:\n {x_treinamento_transformado[0]}')
```

```python
#Treinando há LinearRegression e analisando os coeficientes e o intercepto.

from sklearn.linear_model import LinearRegression

#Instanciando o objeto linearRegression
model = LinearRegression()

#Treinando o modelo usandos os dados de treinamento já transformados
model.fit(x_treinamento_transformado,y_treinamento)

#coefiente e intercepto
print('Coeficientes: ', model.coef_)
print('Intercepto: ', model.intercept_)
```

```python
#Cálculando o RMSE e R² nos conjuntos de treino e teste, aplicando as mesmas transformações aos dados de teste.
from sklearn.metrics import root_mean_squared_error, r2_score

#Avaliando o conjunto de dados de treinamento
y_pred_treinamento = model.predict(x_treinamento_transformado)
rmse_treinamento = root_mean_squared_error(y_treinamento, y_pred_treinamento)
r2_treinamento = r2_score(y_treinamento, y_pred_treinamento)

print(rmse_treinamento)
print(r2_treinamento)

#Transformando os dados de teste
tf_num_teste = imp.transform(x_teste[num_col])
tf_num_teste = scaler.transform(tf_num_teste)
tf_cat_teste = ohe.transform(x_teste[cat_col])
x_teste_transformado = np.concatenate((tf_num_teste, tf_cat_teste), axis=1)

#Avaliando o conjunto de dados de teste
y_pred_teste = model.predict(x_teste_transformado)
rmse_teste = root_mean_squared_error(y_teste, y_pred_teste)
r2_teste = r2_score(y_teste, y_pred_teste)

print(rmse_teste)
print(r2_teste)
```

```python
#Criando o num_preprocessing, full_preprocessing e final_pipeline para simplificar o fluxo de trabalho

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LinearRegression


#Criando o pipeline de pré-processamento para as colunas numericas
num_preprocessing = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler())
])

#Criando o Column Transformer
full_preprocessing = ColumnTransformer([
    ('num', num_preprocessing, num_col),
    ('cat', OneHotEncoder(sparse_output=False, drop='first'), cat_col)
])

#Criando o pipeline final
final_pipeline = Pipeline([
    ('preprocessing', full_preprocessing),
    ('model', LinearRegression())
])

```

```python
final_pipeline.fit(x_treinamento, y_treinamento)
print('Treino concluído')
```

```python
#Avaliando o Pipeline. RMSE e R² em treino e teste, 
#mostrando que o pipeline reproduz o fluxo manual com menos código.

#Avaliando o conjunto de dados de treinamento
y_pred_treinamento = final_pipeline.predict(x_treinamento)
rmse_treinamento = root_mean_squared_error(y_treinamento, y_pred_treinamento)
r2_treinamento = r2_score(y_treinamento, y_pred_treinamento)

print(rmse_treinamento)
print(r2_treinamento)


#Avaliando o conjunto de dados de teste
y_pred_teste = final_pipeline.predict(x_teste)
rmse_teste = np.sqrt(root_mean_squared_error(y_teste, y_pred_teste))
r2_teste = r2_score(y_teste, y_pred_teste)

print(rmse_teste)
print(r2_teste)
```

```python
#Validação cruzada. Validação cruzada de 4 dobras com cross_val_score, usando RMSE e R², 
#com impressão dos scores e das médias.

from sklearn.model_selection import cross_val_score

#Validaçao cruzada com RMSE negativo como criterio da pontuaçao
rmse_scores = cross_val_score(final_pipeline, x_treinamento, y_treinamento, cv=4, scoring='neg_root_mean_squared_error')
rmse_scores_abs = np.abs(rmse_scores)
print('RMSE:')
print('Scores',rmse_scores_abs)
print('Media',np.mean(rmse_scores_abs))

#Validaçao cruzada R2 como criterio de pontuacao
r2_scores = cross_val_score(final_pipeline, x_treinamento, y_treinamento, cv=4, scoring='r2')
print('r2:')
print('Scores',r2_scores)
print('Media',np.mean(r2_scores))
```
